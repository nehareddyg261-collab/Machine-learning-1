```python
import pandas as pd

# File name
CSV_PATH = "/content/archive (1) (1).zip"

# Read CSV file
df = pd.read_csv(CSV_PATH)

# 1. Display first 5 rows
print("FIRST 5 ROWS")
print(df.head())

# 2. Display dataset shape
print("\nDATASET SHAPE")
print(df.shape)

# 3. Display column names
print("\nCOLUMN NAMES")
print(df.columns)

# 4. Display data types
print("\nDATA TYPES")
print(df.dtypes)

# 5. Display dataset information
print("\nDATASET INFORMATION")
df.info()

# 6. Display statistical summary
print("\nSTATISTICAL SUMMARY")
print(df.describe(include="all"))

# 7. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# 8. Check duplicate rows
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# 9. Job role distribution
print("\nJOB TITLE DISTRIBUTION")
print(df["job_title"].value_counts().head(10))

# 10. Work mode distribution
print("\nWORK MODE DISTRIBUTION")
print(df["work_mode"].value_counts())

# 11. Salary tier distribution
print("\nSALARY TIER DISTRIBUTION")
print(df["salary_tier"].value_counts())

# 12. Experience tier distribution
print("\nEXPERIENCE TIER DISTRIBUTION")
print(df["experience_tier"].value_counts())

# 13. Top hiring companies
print("\nTOP 10 HIRING COMPANIES")
print(df["company_name"].value_counts().head(10))

# 14. Top job locations
print("\nTOP 10 JOB LOCATIONS")
print(df["location"].value_counts().head(10))

# 15. Fresher-friendly jobs
print("\nFRESHER FRIENDLY JOBS")
print(df["is_fresher_friendly"].value_counts())

# 16. Senior vs Non-Senior jobs
print("\nSENIOR JOBS")
print(df["is_senior"].value_counts())

# 17. Average salary
print("\nAVERAGE SALARY MIDPOINT (LPA)")
print(df["salary_midpoint_lpa"].mean())

# 18. Average company rating
print("\nAVERAGE COMPANY RATING")
print(df["company_rating"].mean())

# 19. Average experience required
print("\nAVERAGE MINIMUM EXPERIENCE")
print(df["experience_min_yrs"].mean())

print("\nAVERAGE MAXIMUM EXPERIENCE")
print(df["experience_max_yrs"].mean())

# 20. Salary negotiable jobs
print("\nSALARY NEGOTIABLE")
print(df["salary_negotiable"].value_counts())
```
